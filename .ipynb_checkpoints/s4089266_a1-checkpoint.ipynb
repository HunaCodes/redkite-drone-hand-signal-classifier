<h1>SETTING UP THE DATAFRAME</h1>

```python
DATASET_ROOT = Path("../data_resized")
print("Path check: ", DATASET_ROOT.exists())
```

Now we should explicilty give each class a index so it does not unexpectedly change if we were to simply rely on
directory ordering

```python
# Define all the 13 different classes 
CLASS_NAMES = [
    "AllClear",
    "HaveCommand",
    "Hover",
    "Land",
    "LandingDirection",
    "MoveAhead",
    "MoveDownward",
    "MoveToLeft",
    "MoveToRight",
    "MoveUpward",
    "NotClear",
    "SlowDown",
    "WaveOff"
]

class_to_index = {name: i for i, name in enumerate(CLASS_NAMES)}
index_to_class = dict(enumerate(CLASS_NAMES))
class_to_index
```

Quickly ensure the classes have been defined according to the data folder:

```python
success_count = 0
for i, class_name in enumerate(CLASS_NAMES):
    class_path = DATASET_ROOT / class_name

    if class_path.exists():
        print(f"{i}: {class_name} exists!")
        success_count += 1
    else:
        print(f"{i}: {class_name} DOES NOT EXIST! ERROR!")
print("SUCCESS" if success_count == 13 else "ERROR")
```

**IMPORTANT DECISION:**
We will create a Table in which every row represents a single gesture instance. 
- Each row will have **person_id**, **instance_id**, **class_label** and ordered **5 frame paths**
- Each frame path represents the separate images that will be fed into the CNN

The <i>person_id</i> will be crucial for data splitting to prevent data leakage since we do not want the **same person** to be in both training and validation/test. This can cause the model to learn **person-specific** features rather than generalising to unseen people.

Since each frame column stores the path to the image; Later a custom PyTorch Dataset will use these paths to load the images. 

```python
# Stores instances of a gesture
gesture_samples = []

for class_name in CLASS_NAMES:
    for person_dir in (DATASET_ROOT / class_name).iterdir():
        if not person_dir.is_dir():
            continue
        for instance_dir in person_dir.iterdir():

            frame_paths = sorted(instance_dir.glob("*.png"))
            
            # .DS_SOTRE files --> ignore
            if not instance_dir.is_dir():
                continue
            # Check if any missing or extra images
            if len(frame_paths) != 5:
                print(f"ERROR! {instance_dir} has {len(frame_paths)} frame")
                # Skip over it to prevent any malformed data from entering the dataframe
                continue

            gesture_samples.append({
                "person_id": int(person_dir.name.removeprefix("S")), 
                "instance_id": int(instance_dir.name.removeprefix("Instance")), 
                "class_name": class_name,
                "class_label": class_to_index[class_name],
                "frame_1": frame_paths[0],
                "frame_2": frame_paths[1],
                "frame_3": frame_paths[2],
                "frame_4": frame_paths[3],
                "frame_5": frame_paths[4]
            })
df = pd.DataFrame(gesture_samples)
df
```

A Standard train-validation split may significantly reduce the amount of data available for model training and limit the model’s ability to learn effectively. This requires appropriate adjustments to a model's weights to ensure minority labels are given approximately equal consideration by the model.

**IMPORTANT CONSIDERATION 1 (Evaluation Metrics):** This suggests the use of **Macro f1-score** over accuracy alone, since a model can simply achieve a good overall performance score by working well only on majority classes. Since Macro f1-score focuses on how well the model performs on minority classes, we get a much more accurate representation of model performance. 

**IMPORTANT CONSIDERATION 2 (Splitting Technqiue):** Given resampling is not ideal due to tight deadline, it is more apporpriate to perform splitting in a manner so that each set, train/validation/test, have a similar proportion of these class samples through each set. <h3>3. Frame Integrity Check</h3>

```python
frames = ["frame_1", "frame_2", "frame_3", "frame_4", "frame_5"]

frame_sizes = []
frame_modes = []
corrputed_frames = []

for frame in frames:
    for frame_path in df[frame]:

        try:
            with Image.open(frame_path) as img: 
                frame_sizes.append(img.size) 
                frame_modes.append(img.mode) 
                img.verify()
        except Exception as e:
            corrupted_frames.append((f"{frame_path}: {str(e)}"))

# If there is more than one unique element in the set, then we have some data inconsistencies
print("Image Sizes:", set(frame_sizes))
print("Image Modes:", set(frame_modes))
print("No Corrupt Images" if len(corrputed_frames) == 0 else "There are Corrupt Images")
```

The above check was done to ensure that all images had the same image size, and the same modes. In additon, as part of the PIL library, we use **.verify()** to confirm no image is broken or corrputed which can lead to training failure. 

Ensuring Image size and mode is essential to ensure **data consistency** because deep learning models expect uniform data and predictable mathematical inputs, otherwise it can distort the models training.

<h1>Data Splitting</h1>

On Demo Day, we know that the model should be able to generalise to a **new participants**. Given the importance of **Group Splitting** identified in EDA, we must now ensure that all instances from the same participant must remain in the same split. 

If participants are randomly split across all splits, then the model can 'cheat' by learning person-specific features during training rather than generalising to the gesture movement itself, giving overly-optimistic performance results on test set. These result may seem good but will underperform on new participants during demo day since **new** participants will be used rather than the ones used to train the model.

In addition to this, we must also perform some sort of stratified splitting due to the **class imbalance** detected during EDA. 

Finally, given the size of our dataset is relatively small, the most ideal solution is carrying out K-Fold cross validation (Reasoning explained in EDA).

In summary, our Goal is two meet three strict requirements identifed in EDA:
1. **Group Splitting** -> Prevent Data Leakage
2. **Stratified Splitting** -> Perserve class proportions
3. **K-Fold Cross Validation** -> Small Dataset (ensure reliable results)

<h3>Initial Split, Reserved Test (20%)</h3>

We first begin doing an initial ~80/20 split, to create a development set and test set. The test set will reserved as the **final unseen** data that will be used to evaluate the final model. The development set will be used for training/validation purposes. 

Unfortunately, scikit-learn does not provide a simple holdout splitter that can both do group splitting and stratify the 13 gesture classes. Therefore we will use <i>StratifiedGroupKFold</i> to generate **one** group-stratified hold out split. 

- <i>groups=["person_id"]</i> will prevent the participant leakage while stratification provided will attempt to perserve class proportions; This meets our first two requirements.

- <i>n_splits=5</i> gives approximately a **20% test portion**. 

Note we are using <i>StratifiedGroupKFold</i>, however, this stage is **NOT** cross-validation, because only one split is taken. The actual cross-validation will be performed later.

```python
from sklearn.model_selection import StratifiedGroupKFold

sgkf = StratifiedGroupKFold(n_splits=5, shuffle=True, random_state=SEED)

dev_idx, test_idx = next(sgkf.split(df, y=df["class_label"], groups=df["person_id"]))

dev_df = df.iloc[dev_idx].reset_index(drop=True)
test_df = df.iloc[test_idx].reset_index(drop=True)

print("Development Set:", len(dev_df), "instances")
print("Tets Set:", len(test_df), "instances")
```

<h4>Leakage Check</h4>
We must perform a strict leakage check to ensure that no participant has been assigned to both splits:

```python
assert set(dev_df["person_id"]).isdisjoint(set(test_df["person_id"]))
print("Leakage Check is Successful")
```

<h4>Class Proportion Check</h4>
We must also analyse if the class proportions have been reasonably perserved

```python
split_proportions = pd.concat([
    dev_df["class_name"].value_counts(normalize=True).rename("Development"),
    test_df["class_name"].value_counts(normalize=True).rename("Test")
], axis=1).fillna(0)

split_proportions
```

Before doing Cross Validation splits on the development set, it would be wise to choose an ideal **number of folds**

This is because the dataset only contains a small number of participants, different values of k can produce very different fold sizes and class coverage. We therefore compare multiple **k** fold values to choose a preferred split that best ensures: 

1. Participant separation
2. Includes all 13 classes if possible 
3. Keeps validation sizes reasonably balanced

This provides us with a much more reliable estimate of performance on unseen participants.

```python
def check_cv(k):
    cv = StratifiedGroupKFold(n_splits=k, shuffle=True, random_state=SEED)
    results = []

    for fold, (train_idx, val_idx) in enumerate(cv.split(dev_df, dev_df["class_label"], dev_df["person_id"]), 1):
        train_df = dev_df.iloc[train_idx]
        val_df = dev_df.iloc[val_idx]

        assert set(train_df["person_id"]).isdisjoint(val_df["person_id"])

        results.append({
            "fold": fold,
            "train_size": len(train_df),
            "val_size": len(val_df),
            "val_people": val_df["person_id"].nunique(),
            "missing_classes": len(set(CLASS_NAMES) - set(val_df["class_name"]))
        })

    return pd.DataFrame(results)

print("3-Fold")
display(check_cv(3))

print("4-Fold")
display(check_cv(4))

print("5-Fold")
display(check_cv(5))
```

Observing results, clearly **3-fold** is the most ideal since all 13 classes are present in every validation fold and validation sizes are well balanced (301, 281, 286).

4-fold is very uneven and 5-fold has only 1 participant for one, and is missing 6 classes. 

Now with the remaining development set, **3-fold StartifiedGroupKFold cross-validation** will be applied to allow for model and hyperparameter selection. 

As mentioned previously, this is preferred over a single fixed validation split because the dataset is small. A single validation split would make model selection depend heavily on just a few participants, so the measured performance could vary a lot depending on who was chosen. 

Cross validation on the other hand will evaluate model configuration across **k=3 different training and validation splits**. The performance from all 3 folds will be aggregated into an overall validation estimate, making the result more stable and less dependent on one specific split. In the end we chose the best model confugurations to then be **re-trained** on the entire development set, so we are able to maximise model learning before final testing. 

Once again, groups=df["person_id"] will ensure that each particiapnt can only be assigned to one split while perserving class proportions.

```python
cv = StratifiedGroupKFold(n_splits=3, shuffle=True, random_state=SEED)

cv_splits = list(cv.split(dev_df, y=dev_df["class_label"], groups=dev_df["person_id"]))

for fold, (train_idx, val_idx) in enumerate(cv_splits, start=1):
    train = set(dev_df.iloc[train_idx]["person_id"])
    val = set(dev_df.iloc[val_idx]["person_id"])

    assert train.isdisjoint(val)

    print(
        f"Fold {fold}: "
        f"Train Size ({len(train_idx)}), "
        f"Validation Size ({len(val_idx)})"
    )
```

<h1>Preprocessing and PyTorch DataLoaders</h1>

<h3>1. Preprocessing</h3>

We only need to perform light preprocessing on these images since we want the model to learn important features efficiently but also perserve the original data as much as possible. As a result we perform only the following transformations:
 - Images are resized to 224x224 which reduces memory and computation costs and matches the standard input size used by many ImageNet-pretrained CNNs.
 - Color Augmentation is applied; Mild ColorJitter is applied to training data to reduce the potential model reliance on background/scenery cues identified in EDA while the gesture itself remains consistent and visible. This is applied across all 5 frames since each represent one entire gesture. 

Note we avoid augmentations like **horizontal splitting** and other **rotations** since they can change the meaning of directional gestures such as <i>MoveToLeft</i> and <i>MoveToRight</i>.

```python
# This is the standard for ImageNet
IMG_SIZE = 224

frame_transform = transforms.Compose([
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.ToTensor()
])

train_jitter = transforms.ColorJitter(
    brightness=0.1,
    contrast=0.1
)

normalize = transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
)
```

<h3>2. PyTorch Dataset</h3>

A custom PyTorch Dataset is needed since a single training sample is not a single frame but a complete gesture made up of multiple frames. 

The Dataset will be designed to load the 5 frame paths stored within each dataframe row, apply the exact same preprocessing on them, perserves their order (CRITICAL), and stacks them into one tensor so the model sees it as a full gesture instance. 

```python
from torch.utils.data import Dataset

class GestureDataset(Dataset):
    def __init__(self, dataframe, augment=False):
        self.df = dataframe.reset_index(drop=True)
        self.augment = augment
        self.frame_cols = ["frame_1", "frame_2", "frame_3", "frame_4", "frame_5"]

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        frames = []

        for frame_col in self.frame_cols:
            image = Image.open(row[frame_col]).convert("RGB")
            frames.append(frame_transform(image))

        frames = torch.stack(frames)

        if self.augment:
            frames = train_jitter(frames)

        frames = normalize(frames)
        label = torch.tensor(row["class_label"], dtype=torch.long)

        return frames, label
```

<h3>3. DataLoader Setup</h3>

We will require a DataLoader that simply batches the 5 frame gesture instances for training and validation. 

A **batch** size will essentially determine how many gesture instances are processed by the CNN before it updates its weights. Mini-batching can introduce small variations between gradient steps, which can help the optimiser escape poor minimas and explore the loss surface more effectively, supporting convergence towards better solutions. 

**batch_size = 32** is the most ideal size given our current GPU, so we will use this size for now.

```python
from torch.utils.data import DataLoader

BATCH_SIZE = 32

def create_loaders(train_df, val_df):
    train_dataset = GestureDataset(train_df, augment=True)
    val_dataset = GestureDataset(val_df, augment=False)

    train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True)
    val_loader = DataLoader(val_dataset, batch_size=BATCH_SIZE, shuffle=False)

    return train_loader, val_loader
```
